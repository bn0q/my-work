```python
score = 55

def get_grade(score):
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

print(f"Your grade is: {get_grade(score)}")
```

```python
print(get_grade(85))
print(get_grade(75))
print(get_grade(65.000000000000000000
                ))
print(get_grade(55))
print(get_grade(45))
```

```python
students = {"Alice": 45, "Bob": 50, "Charlie": 60}

def print_student_grades(students):
    for student, score in students.items():
        print("🤮 ~ student:", student, "score:", score)
        grade = get_grade(score)
        print(f"{student} has a score of {score} and a grade of {grade}")
```

```python
print_student_grades(students)
```
